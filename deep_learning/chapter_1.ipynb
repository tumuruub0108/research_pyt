```python
X = torch.arange(12, dtype=torch.float32).reshape((3, 4))
print(X)
Y = torch.tensor([[2.0, 1, 4, 3], [1, 2, 3, 4], [4, 3, 2, 1]])
print(Y)
```

dim нь tensor-ийн аль хэмжээсийн дагуу үйлдэл хийхийг заана.

2 хэмжээст tensor-ийн хувьд:
| `dim`   | Хэмжээс        | Утга                        |
| ------- | -------------- | --------------------------- |
| `dim=0` | Мөрийн хэмжээ  | Дээрээс доош залгана        |
| `dim=1` | Баганын хэмжээ | Зүүнээс баруун тийш залгана |


```python
# мөрөөр нь нийлүүлэх. when dim=0: (3,4)+(3,4) -> (6,4)
torch.cat((X, Y), dim=0)
```

```python
# баганаар нь нийлүүлэх. When dim=1: (3,4)+(3,4) -> ​(3,8)
torch.cat((X, Y), dim=1)
```

```python
# For each position i, j, if X[i, j] and Y[i, j] are equal,
print(X)
print(Y)
X == Y
```

```python
# Summing all the elements in the tensor
X.sum()
```
