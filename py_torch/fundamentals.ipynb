# Introduction to tensors

## creating to tensors
Pytorch tensors are created using 'torch.Tensor()'

```python
# scalar
scalar = torch.tensor(7)
scalar
```

```python
scalar.ndim
```

```python
# get tensor back as Pythin int
scalar.item()
```

```python
# Vector
vector = torch.tensor([7, 7])
vector
```

```python
# dimension is like number of square brackets
vector.ndim
```

```python
vector.shape
```

```python
# MATRIX
MATRIX = torch.tensor([[7, 8],
                       [9, 10]])
MATRIX
```

```python
MATRIX.ndim
```

```python
MATRIX[1]
```

```python
MATRIX.shape
```

```python
# TENSOR
TENSOR = torch.tensor([[[1, 2, 3],
                        [3, 6, 7],
                        [2, 4, 5]]])
TENSOR
```

```python
TENSOR.ndim
```

```python
TENSOR.shape # what is that mean?
```

| Name | What is it? | Number of dimensions | 	Lower or upper (usually/example) |
|---|---|---|---|
| scalar | 	a single number | 0 | 	Lower (a)|
| vector | 	a number with direction (e.g. wind speed) or list of numbers | 1 | Lower (y)|
| matrix | 	a 2-dimensional array of numbers | 2 | 	Upper (Q)|
| tensor | 	an n-dimensional array of numbers | 	n (0 = scalar, 1 = vector) | Usually Upper (T)|
