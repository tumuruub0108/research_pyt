## Random tensors

Why random tensors?
Random tensors are important becaue the way many neural networks learn is that they start with tensors with tensors full of random numbers and then adjust those random numbers to better represent the data.


`Start with random numbers -> look at data -> update random numbers -> look at data -> update random numbers`

the rand is created random number between from 0 to 1

```python
# create a random tensor of size (3, 4)
"""
(3, 4, 3)
Үүнийг дараах байдлаар төсөөлж болно:
3 ширхэг матриц
Матриц бүр 4 row, 3 column

tensor of size (3, 4)
3 row, 4 column

"""
random_tensor = torch.rand(3, 4)
random_tensor
```

### Matrix multiplication
Two main ways of performing multiplication in neural networks and deep learning

* Element-wise multiplication
* Matrix multiplication (dot product)


There are two main rules that performing matrix multiplication needs to satisfy:
1. The inner dimensions must match:
* (3, 2) @ (3, 2) won't work
* (2, 3) @ (3, 2) will work
* (3, 2) @ (2, 3) will work

2. The resulting matri has the shape of the **outer dimensions**:
* (2, 3) @ (3, 2) -> (2, 2)
* (3, 2) @ (2, 3) -> (3, 3)

```python
torch.matmul(torch.rand(3, 2), torch.rand(3, 2))
```

```python
# Element wise multiplication
print(tensor, "*", tensor)
print(f"Equals: { tensor * tensor}")
```

```python
# Matrix multiplication
torch.matmul(tensor, tensor)
```

### One of the most common erros in deep learning: shape errors

```python
# Shapes for matrix multiplication

tensor_a = torch.tensor([[1, 2],
                         [3, 4],
                         [5, 6]])

tensor_b = torch.tensor([[7, 10],
                         [8, 11],
                         [9, 12]])


torch.mm(tensor_a, tensor_b)
```

```python
tensor_a.shape, tensor_b.shape
```

 To fix our tensor shape issues, we can manipulate the shape of one of our tensors using **transpose**.

 A **transpose** switches the axex or dimensions of a given tensor

```python
tensor_b, tensor_b.shape
```

```python
tensor_b.T, tensor_b.T.shape
```

```python
# The matrix multiplication operation works when tensor_b is transposed
print(f"Original shape: tensor_A={tensor_a.shape}, tensor_b = {tensor_b.shape}")
print(f"New shapes: tensor_A={tensor_a.shape}, tensor_b.T = {tensor_b.T.shape}")
print(f"multiplying:{tensor_a.shape} @ {tensor_b.T.shape} <- inner dimensions must match")
print("Output:\n")
output = torch.mm(tensor_a, tensor_b.T)
print(output)
print(f"\nOutput shape: {output.shape}")
```
